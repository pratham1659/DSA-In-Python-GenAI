# Interactive Data Visualization with Plotly

**Note:** This notebook originally used `cufflinks`, but that library is now deprecated. We'll use modern **Plotly Express** instead, which provides the same functionality with better performance and active maintenance.

`plotly` is a library for building interactive charts where you can zoom, pan, and hover over points to see values. Plotly Express makes it easy to create these interactive visualizations directly from pandas DataFrames.

**Usecase:** Create interactive charts that allow users to explore data dynamically without needing separate tools.

This cell imports the libraries needed for the rest of the notebook. `seaborn` loads sample datasets, `pandas` handles data manipulation, and `plotly.express` provides easy-to-use functions for creating interactive plots directly from pandas DataFrames.

**Usecase:** These imports set up the tools needed to load the `tips` dataset and create interactive plotly charts.

```python
import seaborn as sns
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Display plots inline in notebook
import plotly.offline as pyo
pyo.init_notebook_mode(connected=True)
```

`sns.load_dataset('tips')` loads a small built-in sample dataset from seaborn called `tips`, which has one row per restaurant visit with columns like `total_bill`, `tip`, `sex`, `smoker`, `day`, `time`, and `size`.

**Usecase:** Get a real dataset to work with. The DataFrame is stored in the `tips` variable for plotting in the following cells.

```python
tips = sns.load_dataset('tips')
print(f"Dataset shape: {tips.shape}")
tips.head()
```

`px.line()` creates an interactive line chart using Plotly Express. This replaces the old `tips['total_bill'].iplot()` method from cufflinks. You can hover over points to see exact values, zoom in, and pan around.

**Usecase:** Visualize the total bill amount for every row as an interactive line chart.

```python
# Interactive line plot of total_bill (replaces tips['total_bill'].iplot())
fig = px.line(
    x=range(len(tips)), 
    y=tips['total_bill'],
    title='Total Bill Amount Over Records',
    labels={'x': 'Record Number', 'y': 'Total Bill ($)'}
)
fig.show()
```

`px.bar()` creates an interactive bar chart. This replaces the old `tips.groupby('day')['tip'].mean().iplot(kind='bar')` method from cufflinks.

**Usecase:** Compare the average tip amount across different days (Thu, Fri, Sat, Sun) as an interactive bar chart.

```python
# Interactive bar chart of average tip by day (replaces cufflinks version)
avg_tip_by_day = tips.groupby('day')['tip'].mean().reset_index()
fig = px.bar(
    avg_tip_by_day, 
    x='day', 
    y='tip',
    title='Average Tip by Day of Week',
    labels={'tip': 'Average Tip ($)', 'day': 'Day of Week'}
)
fig.show()
```

`px.scatter()` creates an interactive scatter plot. This replaces the old `tips.iplot(kind='scatter', x='total_bill', y='tip', mode='markers')` method from cufflinks.

**Usecase:** Plot `total_bill` against `tip` for every row to see if bigger bills tend to come with bigger tips. Hover over any dot to see exact values.

```python
# Interactive scatter plot (replaces cufflinks version)
fig = px.scatter(
    tips, 
    x='total_bill', 
    y='tip',
    title='Total Bill vs Tip Amount',
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'}
)
fig.show()
```

`px.box()` creates an interactive box plot. This replaces the old `tips[['total_bill','day']].iplot(kind='box', x='total_bill', y='day')` method from cufflinks.

**Usecase:** Compare how `total_bill` is distributed across each `day`, showing median, quartiles, and outliers for each day.

```python
# Interactive box plot (replaces cufflinks version)
fig = px.box(
    tips, 
    x='day', 
    y='total_bill',
    title='Total Bill Distribution by Day',
    labels={'total_bill': 'Total Bill ($)', 'day': 'Day of Week'}
)
fig.show()
```

## Additional Interactive Visualizations

Let's explore some additional interactive plots that Plotly Express makes easy to create, going beyond what cufflinks offered.

```python
# Scatter plot with color coding by smoker status
fig = px.scatter(
    tips, 
    x='total_bill', 
    y='tip', 
    color='smoker',
    title='Total Bill vs Tip (Colored by Smoker Status)',
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'}
)
fig.show()
```

```python
# Interactive histogram
fig = px.histogram(
    tips, 
    x='total_bill', 
    nbins=20,
    title='Distribution of Total Bills',
    labels={'total_bill': 'Total Bill ($)', 'count': 'Frequency'}
)
fig.show()
```

```python
# Interactive violin plot
fig = px.violin(
    tips, 
    x='day', 
    y='tip',
    title='Tip Distribution by Day (Violin Plot)',
    labels={'tip': 'Tip ($)', 'day': 'Day of Week'}
)
fig.show()
```

```python
# 3D scatter plot (something cufflinks couldn't do easily)
fig = px.scatter_3d(
    tips, 
    x='total_bill', 
    y='tip', 
    z='size',
    color='time',
    title='3D View: Total Bill vs Tip vs Party Size',
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)', 'size': 'Party Size'}
)
fig.show()
```

## Summary

This notebook demonstrates modern interactive data visualization using **Plotly Express** as a replacement for the deprecated **cufflinks** library. 

### Key Advantages of Plotly Express over Cufflinks:
- **Actively maintained** (cufflinks is deprecated)
- **Better performance** and fewer dependencies
- **More plot types** available (3D plots, animated plots, etc.)
- **Consistent API** that's easy to learn
- **Better customization** options

### Migration Guide:
- `df.iplot()` → `px.line(df)`
- `df.iplot(kind='bar')` → `px.bar(df)`
- `df.iplot(kind='scatter')` → `px.scatter(df)`
- `df.iplot(kind='box')` → `px.box(df)`

All plots are fully interactive with hover tooltips, zooming, panning, and selection capabilities!
